```python
import pandas as pd
pd.set_option('display.max_columns', 20)
df = pd.read_csv("datasets/monthly_top.csv")
artists = df['artists'].unique()
loa = artists.tolist()
lola = [artist.split(', ') for artist in artists]
looa = [[artist[0]] if len(artist)>1 else artist for artist in lola]
los = [artist[0] for artist in looa]
loua = set(los)
louas = sorted(list(loua))
print(len(louas))
```

```python
import json
import os
import musicbrainzngs as mb


mb.set_useragent("CS889 Course Project - Global Music Consumption", "0.7.1", "brigitte.leewy@gmail.com")
mb.set_rate_limit(limit_or_interval=2.5, new_requests=1)

# loads file for known/unknown artists
known_artists = 'artist_data.json'
unknown_artists = 'unknown_artists.txt'

if os.path.exists(known_artists):
    with open(known_artists, 'r') as f:
        try:
            doac = json.load(f)
        except json.JSONDecodeError:
            doac = {}
else:
    doac = {}

if os.path.exists(unknown_artists):
    with open('unknown_artists.txt', 'r') as f:
        not_found = [line.strip() for line in f.readlines()]
else:
    with open('unknown_artists.txt', 'w') as f:
        not_found = [line.strip() for line in f.readlines()]

# asks musicbrainz for country of origin 
for artist in louas:
    if artist in doac or artist in not_found:
        continue
    else: 
        print(artist)
        search = mb.search_artists(artist)
        print(search)
        if ((len(search['artist-list']) != 0 and search['artist-list'][0]) and search['artist-list'][0].get('country')):
            doac[artist] = search['artist-list'][0]['country']
            with open('artist_data.json', 'w') as f:
                    json.dump(doac, f, indent=4)
        else:
            not_found.append(artist)
            with open('unknown_artists.txt', 'a') as f:
                f.write(artist + '\n')

print(doac)
print(not_found)
```

```python
# loads in country of origin and isolates for first artist (if more than one) and saves file

import json 

with open ('artist_data.json', 'r') as f:
    data = json.load(f)
artist_countries = pd.Series(data).to_frame(name = 'origin_country').reset_index()
artist_countries = artist_countries.rename(columns = {'index': 'artists'})
y = artist_countries['artists'].unique().tolist()
print(len(y))

df['artists'] = df['artists'].str.split(',').str[0]
x = df['artists'].unique().tolist()
print(x)
print(len(x))

df_countries = pd.merge(df, artist_countries, on='artists', how='outer')
df_countries = df_countries.sort_values(['month', 'country', 'borda_score'], ascending=[True, True, False])

df_countries.to_csv("df_country_origin.csv")
```

```python
# pulls out artists that musicbrainz couldn't find country of origin for

x = df_countries[df_countries['origin_country'].isna()]
x1 = x.drop_duplicates(subset='artists', keep='first', inplace=False)

print(x1)
x1.to_csv('artists_missing_country.csv')
```

```python
df_mc = pd.read_csv("datasets/artists_missing_country.csv")

# create dictionary and fill in missing countries of origin
country_map = df_mc.set_index('artists')['origin_country']
df_countries['origin_country'] = df_countries['origin_country'].fillna(df_countries['artists'].map(country_map))
df_countries = df_countries.sort_values(['month', 'country', 'borda_score'], ascending=[True, True, False])

df_countries
df_countries.to_csv("complete_dataset_single_artist.csv")
```
