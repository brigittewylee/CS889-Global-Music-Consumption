```python
import pandas as pd
pd.set_option('display.max_columns', 20)
df = pd.read_csv("datasets/monthly_top.csv")
artists = df['artists'].unique()
loa = artists.tolist()
lola = [artist.split(', ') for artist in artists]
looa = [[artist[0]] if len(artist)>1 else artist for artist in lola]
los = [artist[0] for artist in looa]
loua = set(los)
louas = sorted(list(loua))
print(louas)
```

```python
import json
import os
import time
import musicbrainzngs as mb

mb.set_useragent("CS889 Course Project - Global Music Consumption", "0.1", "brigitte.leewy@gmail.com")
mb.set_rate_limit(limit_or_interval=1.0, new_requests=1)

# loads file for known/unknown artists
known_artists = 'artist_data.json'
unknown_artists = 'unknown_artists.txt'

if os.path.exists(known_artists):
    with open(known_artists, 'r') as f:
        try:
            doac = json.load(f)
        except json.JSONDecodeError:
            doac = {}
else:
    doac = {}

if os.path.exists(unknown_artists):
    with open(unknown_artists, 'r') as f:
        not_found = [line.strip() for line in f.readlines()]

for artist in louas[100:]:
    if artist in doac or artist in not_found:
        continue
    else: 
        print(artist)
        search = mb.search_artists(artist)
        print(search)
        if ((len(search['artist-list']) != 0 and search['artist-list'][0]) and search['artist-list'][0].get('country')):
            doac[artist] = search['artist-list'][0]['country']
            with open('artist_data.json', 'w') as f:
                    json.dump(doac, f, indent=4)
        else:
            not_found.append(artist)

print(doac)
print(not_found)
```
